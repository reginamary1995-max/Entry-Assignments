# Final Insights and Conclusion

## Content Performance
The analysis compared engagement across different post types and content
categories using likes, comments, shares, and the calculated engagement score.
The groupby analysis identifies the post types and categories with the
highest average engagement.

## User Trends
Age-group analysis was performed to understand how engagement changes across
different age groups. Verified and non-verified accounts were also compared
using average engagement, likes, comments, shares, and engagement rate.

## Behavioral Insights
The analysis examined impressions across different times of the day and
watch time across device types. This helps identify periods and devices
associated with higher user activity.

## Sentiment Analysis
Posts were grouped into positive, neutral, and negative sentiment categories.
Engagement metrics were compared to understand how sentiment relates to
content performance.

## Overall Conclusion
The analysis demonstrates how Python, NumPy, Pandas, Matplotlib, Seaborn,
and Plotly can be used to clean, transform, analyze, visualize, and interpret
social media engagement data. The findings can help identify patterns in
content performance, audience behavior, device usage, timing, and sentiment.

```python
# ==========================================
# FINAL INSIGHTS
# ==========================================

# 1. Highest engagement post type
post_type_result = (
    df.groupby('post_type')['engagement_score']
    .mean()
    .sort_values(ascending=False)
)

# 2. Highest engagement category
category_result = (
    df.groupby('post_category')['engagement_score']
    .mean()
    .sort_values(ascending=False)
)

# 3. Highest engagement rate country
country_result = (
    df.groupby('country')['engagement_rate']
    .mean()
    .sort_values(ascending=False)
)

# 4. Highest engagement age group
age_result = (
    df.groupby('age_group', observed=True)['engagement_score']
    .mean()
    .sort_values(ascending=False)
)

# 5. Verified vs non-verified
verified_result = (
    df.groupby('is_verified')['engagement_rate']
    .mean()
)

# 6. Device with highest watch time
device_result = (
    df.groupby('device_type')['watch_time_sec']
    .mean()
    .sort_values(ascending=False)
)

# 7. Highest engagement sentiment
sentiment_result = (
    df.groupby('sentiment')['engagement_score']
    .mean()
    .sort_values(ascending=False)
)


print("=" * 60)
print("FINAL INSIGHTS")
print("=" * 60)

print(
    f"\n1. Highest engagement post type: "
    f"{post_type_result.index[0]}"
)

print(
    f"   Average engagement score: "
    f"{post_type_result.iloc[0]:.2f}"
)

print(
    f"\n2. Highest-performing category: "
    f"{category_result.index[0]}"
)

print(
    f"   Average engagement: "
    f"{category_result.iloc[0]:.2f}"
)

print(
    f"\n3. Country with highest average engagement rate: "
    f"{country_result.index[0]}"
)

print(
    f"   Average engagement rate: "
    f"{country_result.iloc[0]:.2f}"
)

print(
    f"\n4. Age group with highest engagement: "
    f"{age_result.index[0]}"
)

print(
    f"   Average engagement: "
    f"{age_result.iloc[0]:.2f}"
)

print("\n5. Verified vs Non-verified:")

for group, value in verified_result.items():
    print(f"   {group}: {value:.2f}")

print(
    f"\n6. Device with highest average watch time: "
    f"{device_result.index[0]}"
)

print(
    f"   Average watch time: "
    f"{device_result.iloc[0]:.2f}"
)

print(
    f"\n7. Sentiment with highest engagement: "
    f"{sentiment_result.index[0]}"
)

print(
    f"   Average engagement: "
    f"{sentiment_result.iloc[0]:.2f}"
)
```

## Sentiment Analysis

**Negative** sentiment posts showed the highest average engagement, with an average engagement score of **12,731.50**.

Negative and neutral sentiment posts were compared using likes, comments, shares, engagement rate, and engagement score to understand their performance patterns.

## Overall Conclusion

The analysis demonstrates how Python, NumPy, Pandas, Matplotlib, Seaborn, and Plotly can be used to clean, transform, analyze, and visualize social media engagement data.
